## Ingestion Configuration

```python
#Idea is to have a config file : A list of dictionaries where key,value pair in each dictionary is source name, source file path, destination table name.

INGESTION_CONFIG=[
    {
        "source" : "CRM",
        "path" : "/Volumes/practice_db/db_sample_proj/inbound_csv/source_crm/cust_info.csv",
        "table" : "crm_cust_info"
    },
    {
        "source" : "CRM",
        "path" : "/Volumes/practice_db/db_sample_proj/inbound_csv/source_crm/prd_info.csv",
        "table" : "crm_prd_info" 
    },
    {
        "source" : "CRM",
        "path" : "/Volumes/practice_db/db_sample_proj/inbound_csv/source_crm/sales_details.csv",
        "table" : "crm_sales_details" 
    },
    {
        "source" : "ERP",
        "path" : "/Volumes/practice_db/db_sample_proj/inbound_csv/source_erp/CUST_AZ12.csv",
        "table" : "erp_cust_az12" 
    },
    {
        "source" : "ERP",
        "path" : "/Volumes/practice_db/db_sample_proj/inbound_csv/source_erp/LOC_A101.csv",
        "table" : "erp_loc_a101" 
    },
    {
        "source" : "ERP",
        "path" : "/Volumes/practice_db/db_sample_proj/inbound_csv/source_erp/PX_CAT_G1V2.csv",
        "table" : "erm_px_cat_g1v2" 
    }
]
```

## Ingesting Files into Bronze Tables

```python
#Now with the help of config file we gonna automate the loading process in bronze tables
for item in INGESTION_CONFIG:
    print(f"Ingesting {item['source']} -> practice_db.bronze.{item['table']}")

    df= (spark.read.option("header", "true")
         .option("inferSchema","true")
         .csv(item["path"])
    )

    (
        df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable(f"practice_db.bronze.{item['table']}")
    )
```
